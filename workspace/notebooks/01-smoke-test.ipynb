# OpenSysML Jupyter kernel smoke test

Select **SysML v2 (OpenSysML)** as the notebook kernel, then run the cells in order.
This intentionally exercises only a small declaration and a value expression.

```sysml
package Prototype {
  private import ScalarValues::*;

  part def Battery {
    attribute capacity : Real;
  }

  part battery : Battery {
    attribute :>> capacity = 42.0;
  }
}
```
